### Data Preprocess

```python
# Remove unwanted columns

# 'Ticket ID', 'Customer Name', 'Customer Email', 'Customer Age', 'Customer Gender', 'Date of Purchase', 
# 'Ticket Status', 'Resolution', 'Ticket Channel', 'First Response Time', 'Time to Resolution', 
# 'Customer Satisfaction Rating'

df.drop(columns=['Ticket ID', 'Customer Name', 'Customer Email', 'Customer Age', 'Customer Gender', 
                          'Date of Purchase', 'Ticket Status', 'Resolution', 'Ticket Channel', 'First Response Time', 
                          'Time to Resolution', 'Customer Satisfaction Rating', "Product Purchased"], inplace=True)


```

```python
df.head()
```

```python
ticket_des = []

for i in range(8469):
    text = df["Ticket Description"][i]
    text = text.replace("{product_purchased}","product")
    ticket_des.append(text)
```

```python
df["Ticket Description"] = ticket_des

df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
ticket_type = df["Ticket Type"]
df.drop(columns="Ticket Type", inplace=True)
df["Ticket Type"] = ticket_type

df
```

```python
ticket = (df["Ticket Subject"] + ":" + df["Ticket Description"]).tolist()
```

```python
df.drop(columns=["Ticket Subject", "Ticket Description"], inplace=True)
```

```python
df["Ticket"] = ticket
```

```python
df
```

```python
df_tick = df["Ticket Type"]
df_par_tick = df["Ticket Priority"]
df = df.drop(columns=["Ticket Type", "Ticket Priority"])
df["Ticket Type"] = df_tick
df["Ticket Priority"] = df_par_tick

df
```
